```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt

from naloga3 import kmeans

# 1. Generate test image with 2 visually similar red regions
test_img = np.ones((200, 200, 3), dtype=np.uint8) * 255  # white background

# First red rectangle (bright red)
cv.rectangle(test_img, (20, 20), (90, 90), (0, 0, 255), -1)

# Second red rectangle (slightly darker red)
cv.rectangle(test_img, (110, 110), (180, 180), (0, 0, 200), -1)

# 2. Run your custom K-means (assumes kmeans() is already defined in notebook)
result_kmeans = kmeans(test_img, k=3, iteracije=10)

# 3. Show input and result side-by-side
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(test_img, cv.COLOR_BGR2RGB))
plt.title("Input Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(result_kmeans, cv.COLOR_BGR2RGB))
plt.title("K-means Result")
plt.axis("off")

plt.tight_layout()
plt.show()

# 4. Optional: Save result
cv.imwrite("kmeans_result_test_image.png", result_kmeans)
```
